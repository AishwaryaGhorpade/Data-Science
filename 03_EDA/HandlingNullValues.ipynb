# Handling Missing Values

1. **Deleting colum**n:if colum have more missing values and that column is not defendent to other column then delete that column is the best way of handling null values.
dropna(axis=1) helps to delete null values column

problem:this is not best way if column have less null values,because if we delte entire data ,we can miss the valuable information from the column.

# Filling the missing values-Imputation

In this case, we will be filling the missing values with a certain number.

The possible ways to do this are:

Filling the missing data with the mean or median value if it's a numerical variable.

Filling the missing data with mode if it's a categorical value.

Filling the numerical value with 0 or -999, or some other number that will not occur in the data. This can be done so that the machine can recognize that the data is not real or is different.

Filling the categorical value with a new type for the missing values.

for numerical values, we can impute with mean and median
for catgorical data we can impute with mode

```python
df['Item_Weight'].mean()
```

```python
df['Item_Weight'].median()
```

Fiiling with **mean** values,
when we have less ouliers then its better to use **mean**

```python
fill_values_df=df
fill_values_df['Item_Weight']=fill_values_df['Item_Weight'].fillna(df['Item_Weight'].mean())
```

```python
fill_values_df.info()
```

filling with **median**,
when we have more outliers then its better way to fill **meadian**

```python
fill_median_df=df
fill_median_df['Item_Weight']=fill_median_df['Item_Weight'].fillna(df['IItem_Weight'].median())
```

```python
fill_median_df.info()
```

# Forward and backword filling -Imputation

1.Forward Fiiling using ffill

```python
forward_filling_df=df
forward_filling_df['Outlet_Size']=forward_filling_df['Outlet_Size'].ffill()
```

```python
forward_filling_df.info()
```

2.backward filling using bfill

```python
backward_filling_df=df
backward_filling_df['Outlet_Size']=backward_filling_df['Outlet_Size'].bfill()
```
