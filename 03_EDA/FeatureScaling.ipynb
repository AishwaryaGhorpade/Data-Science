```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv('/content/Students Annual Score in History geography subject.csv')
```

```python
df.head()
```

```python
df.describe().round(2)
```

**Task 1:Normalization**

```python
new_df=pd.DataFrame(df,columns=['Test_01','Success'])
```

```python
new_df.head(5)
```

fiiling mean in null value place

```python
new_df['Success']=new_df['Success'].fillna(new_df['Success'].mean())
```

```python
new_df.info()
```

```python
scallar=MinMaxScaler() #Instantiating the MinMaxScaller function
normalized_df=scallar.fit_transform(new_df)
```

```python
print(normalized_df)
```

Let's take small Example

```python
arr=np.array([[2],[3],[5],[6],[6]])
scallar=MinMaxScaler() #Instantiating the MinMaxScaller function
normalized_arr=scallar.fit_transform(arr)
print(normalized_arr)
```

**Task 2:Standardization**

```python
scallar=StandardScaler() #Instantiating the StandardScaler function
Standardized_df=scallar.fit_transform(new_df)
print(Standardized_df)
```
