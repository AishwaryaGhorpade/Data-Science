#Stock Price Prediction using Keras Long Short Term  Memory

```python
#import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset_train=pd.read_csv('/content/Stock_prediction_RNN.csv')
```

```python
dataset_train.head()
```

```python
training_set=dataset_train.iloc[:,1:2].values
```

```python
training_set.shape
```

```python
#feature scalling
from sklearn.preprocessing import MinMaxScaler
sc=MinMaxScaler(feature_range=(0,1))
training_set_scaled=sc.fit_transform(training_set)
```

```python
training_set_scaled.shape
```

```python
#creating a data structure with 60 timestamps and 1 output
x_train=[]
y_train=[]
for i in range(60,248):
  x_train.append(training_set_scaled[i-60:i,0])
  y_train.append(training_set_scaled[i,0])
x_train,y_train=np.array(x_train),np.array(y_train)
```

```python
dataset_train.shape
```

```python
#Reshaping
x_train=np.reshape(x_train,(x_train.shape[0],x_train.shape[1],1))
```

```python
x_train.shape
```

```python
#Part-2 -Building an RNN
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import LSTM
from keras.layers import Dropout
```

```python
input_shape=(x_train.shape[1],1)
```

```python
input_shape
```

```python
#initialising RNN
regressor=Sequential()
# units=50,100,200,256,512,1024
# adding first LSTM layer and some dropout regularisation
regressor.add(LSTM(units=50,return_sequences=True,input_shape=(x_train.shape[1],1)))
regressor.add(Dropout(0.2))
```

```python
# adding second LSTM layer and some dropout regularisation
regressor.add(LSTM(units=50,return_sequences=True))
regressor.add(Dropout(0.2))
```

```python
# adding third LSTM layer and some dropout regularisation
regressor.add(LSTM(units=50,return_sequences=True))
regressor.add(Dropout(0.2))
```

```python
# adding final LSTM layer and some dropout regularisation
regressor.add(LSTM(units=50))
# regressor.add(Dropout(0.2))
```

```python
#adding the output layer
regressor.add(Dense(units=1))

#compile the RNN
regressor.compile(optimizer='rmsprop',loss='mean_squared_error')

#fitting the RNN to the training set
regressor.fit(x_train,y_train,epochs=10,batch_size=32)
```
