```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/Churn_Modelling.csv')
```

```python
df.head()
```

```python
X=df.iloc[:,3:13].values
y=df.iloc[:,13].values
```

```python
X
```

```python
y
```

```python
#encoding categorical data
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
LabelEncoder_X_1=LabelEncoder()
X[:,1]=LabelEncoder_X_1.fit_transform(X[:,1])
LabelEncoder_X_2=LabelEncoder()
X[:,2]=LabelEncoder_X_2.fit_transform(X[:,2])
```

```python
X
```

```python
#splitting the dataset into training and testing
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
#feature scalling
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
#classical ML
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier()
clf.fit(X_train,y_train)
pred=clf.predict(X_test)
```

```python
#making confusion matrix
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,pred)
print(cm)
```

```python
#accuracy score
from sklearn.metrics import accuracy_score
score=accuracy_score(y_test,pred)
print(score*100)
```

#Accuracy from decision Tree is 78.9

```python
#Random forest
from sklearn.ensemble import RandomForestClassifier
clf2=RandomForestClassifier(n_estimators=200)
clf2.fit(X_train,y_train)
pred2=clf2.predict(X_test)
```

```python
#making confusion matrix
from sklearn.metrics import confusion_matrix
cm2=confusion_matrix(y_test,pred2)
print(cm2)
```

```python
#accuracy score
from sklearn.metrics import accuracy_score
score2=accuracy_score(y_test,pred2)
print(score2*100)
```

#Accuracy from Random forest is 86.3

```python
#ANN
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
```

```python
#initializing ANN
clasifier=Sequential()
#units=input+output/2
# initilizer=how ur weights are updated
# relu activation function
# input_dim=input shape(how many feature do u have in the data)

#adding input layer and first hidden layer
clasifier.add(Dense(units=6,kernel_initializer='uniform',activation='relu',input_dim=10))
#adding second hidden layer
clasifier.add(Dense(units=6,kernel_initializer='uniform',activation='relu'))
#adding third hidden layer
clasifier.add(Dense(units=6,kernel_initializer='uniform',activation='relu'))
#adding output layer
clasifier.add(Dense(units=1,kernel_initializer='uniform',activation='sigmoid'))
```

```python
#compile ANN
clasifier.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
y_train[0]
#fitting the ANN training to Training set
clasifier.fit(X_train,y_train,batch_size=10,epochs=50)
```

```python
#predicting the model
y_pred_ann=clasifier.predict(X_test)
y_pred_ann=(y_pred_ann>0.5)
cm3=confusion_matrix(y_test,y_pred_ann)
print(cm3)
```

```python
#accuracy
score3=accuracy_score(y_test,y_pred_ann)
print(score3*100)
```

#Accuracy for ANN is 84.2
