```python
import numpy as np
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import random
from skimage import exposure
from skimage.util import random_noise
from skimage import transform
from cv2 import resize
```

```python
img=mpimg.imread('/content/cat.jpg')
```

```python
plt.imshow(img)
```

#RESCALE

```python
img_rescale=resize(img,(150,150))
```

```python
plt.imshow(img_rescale)
```

```python
img_rescale2=resize(img,(50,50))
```

```python
plt.imshow(img_rescale2)
```

#FLIPPING

```python
horizontal_flip=np.fliplr(img)
```

```python
plt.imshow(horizontal_flip)
```

```python
verticle_flip=np.flipud(img)
```

```python
plt.imshow(verticle_flip)
```

#ROTATION

```python
trans_img=transform.rotate(img,90)
#90deg rotate
```

```python
plt.imshow(trans_img)
```

```python
trans_img2=transform.rotate(img,random.uniform(-40,40))
plt.imshow(trans_img2)
```

#adding noise

```python
img_noise=random_noise(img,mode='s&p',clip=True)
plt.imshow(img_noise)
```
