```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    'Year':[1920,1930,1940,1950,1960,1970,1980,1990,2000,2010,2020],
    'Exchange Rate':[65,69,71,64,62,59,72,71,75,78,81]
}
```

for reading csv excel file use pandas method,pd.read_csv(file.csv),pd.read_excel(file.exel)

```python
#converting data in dataframe so that easy to use numpy and pandas methods
df=pd.DataFrame(data)
#or we can give columnls as here it unserstands keys are columns
# df=pd.DataFrame(data,columns=['Year','Exchange Rate'])
```

```python
df
```

```python
type(df)
```

```python
df.plot(x='Year',y="Exchange Rate",kind="bar")
plt.show()
```

```python
df.plot(x='Year',y="Exchange Rate",kind="line")
plt.show()
```

```python
df.plot(x='Year',y="Exchange Rate",kind="area")
plt.show()
```

```python
#this is using matplotlib libraries
x=np.array([10,20,30,40,50])
y=np.array([600,50,1000,400,300])
plt.scatter(x,y)
plt.show()
```

```python
data={
    'Tasks':[100,200,300]
}
df=pd.DataFrame(data,columns=["Tasks"],index=['Pending','Completed','Ongoing'])
df.plot.pie(y='Tasks',figsize=(5,5))
plt.show()
```

```python
df
```

```python
# without index ,it takes 0,1,2 etc
df=pd.DataFrame(data,columns=["Tasks"])
df.plot.pie(y='Tasks',figsize=(5,5))
plt.show()
```

```python
labels_ex='JavaScript','Java','Python','R'
sizes=[15,30,45,10]
explode_labels=(0,0.2,0,0)
fig1,ax1=plt.subplots() #will get 0.2 expode Java part in pie chart
ax1.pie(sizes,explode=explode_labels,labels=labels_ex,shadow=True,startangle=90)
ax1.axis('equal')
plt.show()
```

```python
labels_ex='JavaScript','Java','Python','R'
sizes=[15,30,45,10]
explode_labels=(0,0.2,0.3,0)
fig1,ax1=plt.subplots() #will get 0.2 expode Java part in pie chart
ax1.pie(sizes,explode=explode_labels,labels=labels_ex,shadow=True,startangle=90)
ax1.axis('equal')
plt.show()
```
