concat and Merging

```python
df1=pd.DataFrame({
    'A':['A0','A1','A2','A3'],
    'B':['B0','B1','B2','B3'],
    'C':['C0','C1','C2','C3'],
    'D':['D0','D1','D2','D3']
})
df2=pd.DataFrame({
    'E':['A4','A5','A6','A7'],
    'F':['B4','B5','B6','B7'],
    'C':['C0','C3','C6','C7'],
    'G':['D4','D5','D6','D7']
})
```

```python
df1
```

```python
df2
```

```python
merge_df=pd.merge(df1,df2,on='C', how='outer')
```

```python
merge_df
```

```python
merge_df1=pd.merge(df1,df2,on='C', how='inner')
```

```python
merge_df1
```

```python
merge_df2=pd.merge(df1,df2,on='C', how='left')
```

```python
merge_df2
```

```python
merge_df3=pd.merge(df1,df2,on='C', how='right')
```

```python
merge_df3
```
