```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Telco-Customer-Churn.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
#TotalCharges is object so need to convert to numeric
df.TotalCharges=pd.to_numeric(df.TotalCharges,errors='coerce')
```

```python
df.info()
```

```python
df.dropna(how='any',inplace=True)
```

```python
X=df.drop(['customerID','Churn'],axis=1)
y=df.Churn.values
```

```python
X
```

```python
y
```

```python
X.columns
```

```python
#convert categorical feature to numerical -->Feature encoding
X=pd.get_dummies(X,columns=['gender', 'Partner', 'Dependents',
       'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
       'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV',
       'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod'],drop_first=True)
```

```python
X.head(1)
```

```python
#spliting data into training and test
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
#Feature Scalling
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train_sc=sc.fit_transform(X_train)
X_test_sc=sc.transform(X_test)
```

```python
X_train_sc
```

#KNN Classifier

```python
#call the KNN classifier
from sklearn.neighbors import KNeighborsClassifier

#Initializing the classifier
model=KNeighborsClassifier()

#passing the data to model
model.fit(X_train_sc,y_train)
```

```python
y_pred=model.predict(X_test_sc)
```

```python
y_pred
```

```python
y_test
```

```python
#classification metrics
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
#finding accuracy percentage
round(accuracy_score(y_test,y_pred)*100,2)
```

#Decision Tree Classifier

```python
#call the Decision Tree classifier
from sklearn.tree import DecisionTreeClassifier

#Initializing the classifier
model_dt=DecisionTreeClassifier()

#passing the data to model
model_dt.fit(X_train_sc,y_train)
```

```python
y_pred_dt=model_dt.predict(X_test_sc)
```

```python
#classification metrics
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred_dt)
```

```python
#finding accuracy percentage of dt
round(accuracy_score(y_test,y_pred_dt)*100,2)
```

#Visualization of Decision Tree

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(10,10))
plot_tree(model_dt,filled=True,feature_names=X_train.columns,class_names=True)
plt.show()
```

#Random forest

```python
#call the Random Forest classifier
from sklearn.ensemble import RandomForestClassifier

#Initializing the classifier
model_rf=RandomForestClassifier()

#passing the data to model
model_rf.fit(X_train_sc,y_train)
```

```python
y_pred_rf=model_rf.predict(X_test_sc)
```

```python
#finding accuracy percentage of rf
round(accuracy_score(y_test,y_pred_rf)*100,2)
```
