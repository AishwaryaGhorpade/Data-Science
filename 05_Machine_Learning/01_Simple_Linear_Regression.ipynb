```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#read data file
df=pd.read_csv("/content/50_Startups_dataset.csv")
```

```python
#Devide the data into dependent vs independents
X=df.iloc[:,3].values
Y=df.iloc[:,-1].values
```

```python
# #Feature Encoding - OneHotEncoder
# from sklearn.compose import ColumnTransformer
# from sklearn.preprocessing import OneHotEncoder
# ct=ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[3])],remainder='passthrough')
# X=np.array(ct.fit_transform(X))

#Feature encoding not needed here -bcz X have all numeric value so no need to conver to numeric
```

```python
#Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
X_train=X_train.reshape(-1,1)
X_test=X_test.reshape(-1,1)
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
#Create LR model
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(X_train,Y_train)
```

```python
#predictions
Y_pred=regressor.predict(X_test)
```

```python
#ploting data Y_pred vs Y_test
plt.plot(Y_test,color='blue',label='Y_test')
plt.plot(Y_pred,color='red',label='prediction')
plt.show()
```
