```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.metrics import r2_score
```

```python
#data
x=6*np.random.rand(200,1)-3
y=0.8*x**2 + 0.9 * x + 2+ np.random.randn(200,1)
# y=bx**2 +bx + c
```

```python
plt.plot(x,y,'b.')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
#Train test split
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2)
```

#Appling Linear regression model

```python
#Appling Linear regression model
lr=LinearRegression()
lr.fit(X_train,Y_train)
```

```python
Y_pred=lr.predict(X_test)
r2_score_lr=r2_score(Y_pred,Y_test)
print("R2_scrore for lr model:",r2_score_lr)
```

```python
plt.plot(X_train,lr.predict(X_train),color='r')
plt.plot(x,y,'b.')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

#Applying Polynomial Linear Regression

```python
poly=PolynomialFeatures(degree=2,include_bias=True)
x_train_trans=poly.fit_transform(X_train)
x_test_trans=poly.transform(X_test)
```

```python
print(X_train[0])
```

```python
print(x_train_trans[0])
```

```python
lr2=LinearRegression()
lr2.fit(x_train_trans,Y_train)
```

```python
y_pred2=lr2.predict(x_test_trans)
r2_score_poly=r2_score(y_pred2,Y_test)
print("R2_scrore for poly model:",r2_score_poly)
```

```python
print(lr2.coef_)
```

```python
print(lr2.intercept_)
```

```python
x_new=np.linspace(-3,3,200).reshape(200,1)
x_new_poly=poly.transform(x_new)
y_new=lr2.predict(x_new_poly)
```

```python
plt.plot(x_new,y_new,'r-',linewidth=2,label="prediction")
plt.plot(X_train,Y_train,'b.',label="Trainng")
plt.plot(X_test,Y_test,'g.',label="Testing")
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
